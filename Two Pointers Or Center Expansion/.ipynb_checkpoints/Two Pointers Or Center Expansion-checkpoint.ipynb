```python
def oddPalindrome(ind, stringLength):
    i = ind-1
    j = ind+1
    # odd palindrome 
    palindromeLength = 1
    # print(i,j)
    while i>=0 and j<stringLength:
        if strng[i] == strng[j]:
            palindromeLength+=2
            i-=1
            j+=1
        else:
            break
    return palindromeLength
    
def evenPalindrome(ind, stringLength):
    i = ind-1
    j = ind
    # odd palindrome 
    palindromeLength = 0

    while i>=0 and j<stringLength:
        if strng[i] == strng[j]:
            palindromeLength+=2
            i-=1
            j+=1
        else:
            break
    return palindromeLength

    
strng = "forgeeksskeegfor"
n, maxLength = len(strng), 0

for ind in range(1, n):
    maxLength = max(maxLength, oddPalindrome(ind, n), evenPalindrome(ind, n))
print(maxLength)

            
```
