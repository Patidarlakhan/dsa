### Trick 1 : Monotonic Decreasing Stack

#### Question 1: Next Greater Element

##### Trick : Monotonic decreasing stack from right.  

Question 1 : Why from right side?
Ans : so that we can keep track of the monotonic decreasing element from the last element to the current element, means the next big element to the current element will be on the top of the stack.

```python
### Example
nums = [2, 1, 4, 3, 5]
final = [-1]*len(nums)
stack = []
for i in range(len(nums)-1, -1, -1):
  while stack and stack[-1]<=nums[i]:
    stack.pop()
  if stack:
    final[i] = stack[-1]
  stack.append(nums[i])
  
print(final)
```

Question 2: Stock Span Problem

```python
# Example
prices = [100, 80, 60, 70, 60, 75, 85]

indes_stack = []
final = []
for i, price in enumerate(prices):
  while indes_stack and prices[indes_stack[-1]] <= price:
    indes_stack.pop()
  if not indes_stack:
    span = i + 1
  else:
    span = i - indes_stack[-1]
  indes_stack.append(i)
  final.append(span)
print(final)
```

Question 3: Largest Rectangle in Histogram

```python
heights = [2, 1, 5, 6, 2, 3]

n = len(heights)
mx = 0

for i, h in enumerate(heights):
    
  for j in range(i-1, -1, -1):
    if heights[j] < heights[i]:
      mx = max(mx, (i-j)*heights[i])
      break
        
  for j in range(i+1, n):
    if heights[j] < heights[i]:
      mx = max(mx, (j-i)*heights[i])
      break
        
print(mx)
```

Question 3: Remove Adjacent Duplicates

```python
s = "abbaca"

stack = []
for ch in  s:
  if stack and stack[-1]==ch:
    stack.pop()
  else:
    stack.append(ch)
print(''.join(stack))
```

Question 4: Evaluate Postfix Expressions

```python
tokens = ["2", "1", "+", "3", "*"]
stack = []
s = 0
for token in tokens:
    if token not in {"+", "-", "*", "/"}:
        stack.append(int(token))
    else:
        b = stack.pop()
        a = stack.pop()

        if token == "+":
            result = a + b

        elif token == "-":
            result = a - b

        elif token == "*":
            result = a * b

        else:
            result = int(a / b)

        stack.append(result)
print(stack[-1])
```

#### Question 5 : Min Stack

<strong>Problem

Design a stack that supports:

Push<br>
Pop<br>
Top<br>
Get minimum

All in O(1) time.

```python
class Minstack:
    def __init__(self):
        self.stack = []
    def push(self, val):
        if not self.stack:
            self.stack.append((val, val))
        else:
            current_min = min(val, self.stack[-1][1])
            self.stack.append((val, current_min))
    def pop(self):
        return self.stack.pop()[0]
    def top(self):
        return self.stack[-1][0]
    def getMin(self):
        return self.stack[-1][1]

s = Minstack()
ps = [5,3,7,2]
for psh in ps:
    s.push(psh)
print(s.stack)
```

Question 6 : Remove K Digits

<strong>Input

num = "1432219", <br>
k = 3

<strong>Output:</strong><br>
"1219"

```python
nums = "1432219"
k = 3
stack = []
c = 0

for num in nums:
    while stack and int(stack[-1])>int(num) and c < k:
        stack.pop()
        c += 1
    stack.append(num)
    
while c < k:
    stack.pop()
    c += 1
print(''.join(stack))
```
